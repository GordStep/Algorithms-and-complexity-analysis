# Алгоритм поиска подстроки КМП и БМ

```python
def SearchKMP(inp_str, key):
    len_k = len(key)
    sim_chars = 0
    # print(inp_str, key)
    ind = 0
    while ind < (len(inp_str) - len_k + 1):
        flag = True
        for ind_sim in range(len_k):
            # print(ind, ind_sim, inp_str[ind+ind_sim], key[ind_sim])
            if inp_str[ind + ind_sim] != key[ind_sim]:
                sim_chars = ind_sim
                # print('no', inp_str[ind+ind_sim], key[ind_sim], sim_chars)
                flag = False
                break
        if flag:
            return ind
        else:
            ind += sim_chars
        ind += 1
    
    return -1
```

```python
text = "abada bado bc" 
pattern = "bado"

SearchKMP(text, pattern)
```

```python
def SearchBM(inp_str, key):
    len_k = len(key)
    sim_chars = 0
    # print(inp_str, key)
    ind = len(key) - 1
    while ind < (len(inp_str) - len_k + 1):
        flag = True
        for ind_sim in range(len_k):
            # print(ind, len_k - ind_sim - 1, inp_str[ind - ind_sim], key[-ind_sim - 1])
            if inp_str[ind - ind_sim] != key[-ind_sim - 1]:
                sim_chars = ind_sim
                # print('no', inp_str[ind - ind_sim], key[-ind_sim - 1], len_k - sim_chars - 1)
                flag = False
                break
        if flag:
            return ind - len_k + 1
        else:
            ind += sim_chars
        ind += 1
    
    return -1
```
