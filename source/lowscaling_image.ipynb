```python
from PIL import Image, ImageDraw 
```

```python
def LowscalingImage2x(image):
    width = image.size[0] #Определяем ширину. 
    height = image.size[1] #Определяем высоту. 
    
    low_image = Image.new(mode="RGB", size=(width // 2, height // 2))
    draw = ImageDraw.Draw(low_image) #Создаем инструмент для рисования. 
    	
    orig_pix = image.load() #Выгружаем значения пикселей.
    
    for i in range(width // 2):
        for j in range(height // 2):
            
            x = 2 * i
            y = 2 * j
            
            if x == width - 1:
                x = width - 2
            
            if y == height - 1:
                y = height - 2
                
            pixel = orig_pix[x, y]
            
            draw.point((i, j), pixel)
                    
    return low_image
```

```python
image = Image.open("./data/image_U2xS.png") #Открываем изображение. 
image.size
```

```python
l_img = LowscalingImage2x(image)

l_img.show()

l_img.save('./data/image_L2x.png')
```
