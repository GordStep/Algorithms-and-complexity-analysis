# Быстрая сортировка 
Реализовать алгоритмы быстрой сортировки Ломуто, Хоара

Массив на 256 элементов, он должен влючкать цифры, числа, буквы, буквосочетания, слова(латиница)


Сравнить алгоритмы по времени

```python
import random
import numpy as np
import time
```

```python
def Swap(arr, i, j):
    arr[i], arr[j] = arr[j], arr[i]
```

```python
def Partition_Hoa(arr, l, r):
    p = arr[(l + r) // 2]
    i = l
    j = r
    
    while True:
        #print(p)
        while arr[i] <= p:
            i += 1
            #print(f'i = {i}')
        while arr[j] > p:
            j -= 1
            #print(f'j = {j}')
        
        if i >= j:
            return j
    
        Swap(arr, i, j)
        i += 1
        j -= 1 
```

```python
def Partition(arr, l, r):
    p = arr[l]
    i = l
    j = r
    
    while i < j:
        #print(p)
        while i <= r and arr[i] <= p:
            i += 1
            #print(f'i = {i}')
        while j >= l and arr[j] > p:
            j -= 1
            #print(f'j = {j}')
        
        if i < j:
            Swap(arr, i, j)  # просто вызываем Swap без присваивания
    
    Swap(arr, l, j)  # меняем опорный элемент
    return j
    
```

```python
def Partition_Lomuto(arr, l, r):
    pivot = arr[r]
    i = l
    
    for j in range(l, r):
        if arr[j] <= pivot:
            Swap(arr, i, j)
            i += 1
    
    Swap(arr, i, r)
    return i
```

```python
def QuickSort_Lomuto(arr, l, r):
    if l < r:
        s = Partition_Lomuto(arr, l, r)
        QuickSort_Lomuto(arr, l, s-1) 
        QuickSort_Lomuto(arr, s+1, r) 
```

```python
def QuickSort_Hoara(arr, l, r):
    if l < r:
        s = Partition_Hoa(arr, l, r)
        QuickSort_Hoara(arr, l, s-1) 
        QuickSort_Hoara(arr, s+1, r) 
```

```python
def QuickSort(arr, l, r):
    if l < r:
        s = Partition(arr, l, r)
        QuickSort(arr, l, s-1) 
        QuickSort(arr, s+1, r) 
```

```python


# Готовый массив со случайными элементами
array = [
    
    # Буквы
    'A', 'b', 'C', 'd', 'E', 'f', 'G', 'h', 'I', 'j',
    'K', 'l', 'M', 'n', 'O', 'p', 'Q', 'r', 'S', 't',
    'U', 'v', 'W', 'x', 'Y', 'z',
    
        # Буквосочетания
    'abc', 'bcd', 'cde', 'def', 'efg', 'fgh', 'ghi', 'hij', 'ijk', 'jkl',
    'klm', 'lmn', 'mno', 'nop', 'opq', 'pqr', 'qrs', 'rst', 'stu', 'tuv',
    'uvw', 'vwx', 'wxy', 'xyz', 'ab', 'cd', 'ef', 'gh', 'ij', 'kl',
    'mn', 'op', 'qr', 'st', 'uv', 'wx', 'yz', 'abcde', 'fghij', 'klmno', 'pqrst',
    
    # Слова
    'apple', 'banana', 'cherry', 'dog', 'elephant', 'forest', 'garden',
    'happy', 'intelligent', 'joyful', 'kind', 'lovely', 'mountain',
    'nature', 'ocean', 'python', 'quick', 'river', 'sun', 'tree',
    'universe', 'violet', 'wonder', 'xray', 'yellow', 'zebra',
    'code', 'data', 'algorithm', 'function', 'variable'
]

for i in range(100):
    array.append(str(i))

L = 256 * 8
N = 1000
random_mass = [random.choice(array) for _ in range(L)]

random_mass
```

```python
def IsSort(arr):
    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            return False
    return True
```

```python
# Тестируем Ломуто

times_l = []

# Тестируем Хоара
times_h = []
#L = 1_000_00
#random_mass = [random.randint(0, 100) for _ in range(L)]

for _ in range(N):
    random_mass = [random.choice(array) for _ in range(L)]
    
    # Тестирование Хоара
    test_mass = random_mass.copy()
    
    time_start = time.time()
    #print(test_mass)
    QuickSort(test_mass, 0, len(test_mass) - 1)
    #print(test_mass)
    time_end = time.time()
    
    times_h.append(time_end - time_start)
    # print(IsSort(test_mass))

    # Тестирование Ломуто
    test_mass = random_mass.copy()
    
    time_start = time.time()
    #print(test_mass)
    QuickSort_Lomuto(test_mass, 0, len(test_mass) - 1)
    #print(test_mass)
    time_end = time.time()
    
    times_l.append(time_end - time_start)
    # print(IsSort(test_mass))
    
len(times_l), len(times_h)
```

```python
print(f'Результаты {N} тестов:')
print(f'Характеристи массива:\nL = {L} элементов\n')

av_h = np.average(times_h)
av_l = np.average(times_l)

delta = np.abs(av_h - av_l)

print(f'Среднее время выполения Ломуто: {np.average(times_l):05f}')
print(f'Среднее время выполения Хоара:  {np.average(times_h):05f}\n')

faster_n = 'Ломуто'
if av_h < av_l:
    faster_n = 'Хоара'

print(f'Метод разбиения {faster_n} в среднем быстрее:')
    
min_t = min(av_h, av_l)
max_t = max(av_h, av_l)

print(f'на {round(delta, 7)} сек')
print(f'в {round(max_t / min_t, 7)} раза')
print(f'на {round(delta / min_t * 100, 1)}%')
```

```python
import matplotlib.pyplot as plt

t = [*range(N)]

plt.scatter(t, times_h)
plt.scatter(t, times_l)

plt.title('Время сортировки массива (меньше - лучше)')
plt.legend(['Хоар', 'Ломуто'])

plt.xlabel('Номер итерации')
plt.ylabel('Время, сек')
```
