5) Выполнить смещение полосы вверх и окрашивание в инверсию.

```python
from PIL import Image, ImageDraw
```

```python
image_path = 'C:\\Users\\Gordey\\Downloads\\a928209cd53c803870a9961d4727d649.jpg'
image = Image.open(image_path)
```

```python
draw = ImageDraw.Draw(image)

width = image.size[0]
height = image.size[1]

pix = image.load()

width, height
```

```python
image
```

```python
move_pix = 30


orig_pix = image.copy().load() # сохраняем исходное изображение 

down_line = 200
top_line = 400
        
        
for x in range(width):
    for y in range(height):
        
        # Инверсия цвета 
        r = pix[x, y][0]
        g = pix[x, y][1]
        b = pix[x, y][2]
        draw.point((x, y), (255 - r, 255 - g, 255 - b))
        
        # Сдвиг линии
        if down_line - x <= y and y <= top_line - x:            
            draw.point(((x + move_pix) % width, (y - move_pix) % height), orig_pix[x, y])
        
        # ограничивающие прямые 
        if down_line - x == y or y == top_line - x:
            draw.point((y, x), (255, 0, 0))
            


image.show()
image.save("test.jpg")
```
